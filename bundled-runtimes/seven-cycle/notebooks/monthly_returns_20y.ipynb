# 20年间月度回报（复权/指数增长率）

加载 `output/monthly_returns_20y.parquet`，逐分类展示表格与折线图。

```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

ROOT = Path('..').resolve()
df = pd.read_parquet(ROOT/'output/monthly_returns_20y.parquet')
df.tail()
```

```python
sns.set(style='whitegrid')
for cat in df.columns.get_level_values(0).unique():
    sub = df[cat]
    display(sub.tail(36))
    years = sub.index
    fig, ax = plt.subplots(figsize=(12, 6))
    for col in sub.columns:
        ax.plot(years, (sub[col]*100.0), lw=1.0, label=str(col))
    ax.set_title(f"{cat} 月度回报（%）")
    ax.set_ylabel('%')
    ax.axhline(0, color='k', lw=0.8)
    if len(sub.columns) <= 12:
        ax.legend(loc='best', fontsize=9)
    else:
        ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1), fontsize=7)
        fig.subplots_adjust(right=0.78)
    plt.show()
```
